```python
df = pd.read_csv("/content/covid_toy - covid_toy (1).csv")
```

```python
df.head()
```

```python
# Gender ko numerical
df["gender"] = df["gender"].map({
    "Male": 1,
    "Female": 0
})

# Cough ko numerical
df["cough"] = df["cough"].map({
    "Mild": 1,
    "Strong": 2
})

# City ko numerical using One-Hot Encoding
df = pd.get_dummies(
    df,
    columns=["city"],
    dtype=int
)

print(df.head())
```

```python
df['has_covid'] = df['has_covid'].dropna()
```

```python
df['has_covid'].isna().sum()
```

```python
df['has_covid'] = df['has_covid'].map({
    "Yes": 1,
    "No": 0
})
```

```python
df['has_covid']
```

```python
X = df.drop(columns=["has_covid"])
Y = df['has_covid']
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y, test_size= 0.2,random_state= 42)
```

```python
base_learners = []
num_base_learners = 10
```

```python
for i in range(num_base_learners):

    bootstrap_indices = np.random.choice(
        len(X_train),
        size=len(X_train),
        replace=True
    )

    x_bootstrap = X_train.iloc[bootstrap_indices]

    y_bootstrap = Y_train.iloc[bootstrap_indices]
```

```python
from sklearn.ensemble import RandomForestClassifier
import numpy as np

base_learners = []   # <-- important

base_learner = RandomForestClassifier(
    n_estimators=10,
    random_state=42
)

base_learner.fit(x_bootstrap, y_bootstrap)

base_learners.append(base_learner)

base_predictions = []

for base_learner in base_learners:
    y_pred = base_learner.predict(X_test)
    base_predictions.append(y_pred)

ensemble_predictions = np.round(
    np.mean(base_predictions, axis=0)
)

```

```python
accuracy = accuracy_score(Y_test,ensemble_predictions)
print("covid dataset accuracy ---> ", accuracy)
```
