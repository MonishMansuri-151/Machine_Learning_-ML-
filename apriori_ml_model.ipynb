```python
from apyori import apriori

# Data preprocessing
transaction = []

for i in range(7501):
    transaction.append([
        str(df.values[i, j])
        for j in range(20)
    ])

# Training the dataset
rules = apriori(
    transaction,
    min_support=0.003,
    min_confidence=0.2,
    min_lift=3,
    min_length=2
)

# Visualizing the results
a = list(rules)

result = [
    list(a[i][0])
    for i in range(len(a))
]
```
