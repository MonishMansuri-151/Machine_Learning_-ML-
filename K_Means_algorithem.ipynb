```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/mall - mall.csv")
```

```python
df = df.drop(columns=['CustomerID',"Genre"])
```

```python
df.head()
```

```python
x = df.iloc[:,[0,1]].values
```

```python
from sklearn.cluster import KMeans
```

```python
import matplotlib.pyplot as plt
```

```python
a = []
```

```python
for i in range(1,11):
  b = KMeans(n_clusters= i , init= "k-means++",random_state= 42)
  b.fit(x)
  a.append(b.inertia_)
plt.plot(range(1,11),a)
plt.title("The Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS_list")
plt.show()
```

```python
# from the above plot we can see the elbow point is at 4. so the number of cluster here will be 4
```

```python

b = KMeans(n_clusters= 4, init= "k-means++",random_state= 42)
y_predict = b.fit_predict(x)
```

```python
#visulazing the cluster
```

```python
plt.scatter(x[y_predict == 0, 0], x[y_predict == 0, 1],
            s=100, c="blue", label="Cluster1")

plt.scatter(x[y_predict == 1, 0], x[y_predict == 1, 1],
            s=100, c="green", label="Cluster2")

plt.scatter(x[y_predict == 2, 0], x[y_predict == 2, 1],
            s=100, c="red", label="Cluster3")

plt.scatter(x[y_predict == 3, 0], x[y_predict == 3, 1],
            s=100, c="cyan", label="Cluster4")

plt.scatter(b.cluster_centers_[:, 0], b.cluster_centers_[:, 1],
            s=300, c="yellow", label="Centroids")

plt.title("Clusters of Customers")
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.legend()
plt.show()
```
